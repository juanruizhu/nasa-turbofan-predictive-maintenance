```python
import pandas as pd

# 1. Definimos los nombres de las 26 columnas
nombres_columnas = ['id_motor', 'ciclo', 'ajuste_1', 'ajuste_2', 'ajuste_3'] + [f'sensor_{i}' for i in range(1, 22)]

# 2. Leemos el archivo.
# sep='\s+' -> le indicamos que los números están separados por espacios
# header=None -> le indicamos que la primera línea ya tiene datos, no los títulos
df = pd.read_csv("train_FD001.txt", sep=r'\s+', header=None, names=nombres_columnas)

# 3. Mostramos las primeras 5 filas
display(df.head())
```

```python
# 1. CALCULAR LA VIDA ÚTIL RESTANTE (RUL)
# Buscamos cuántos vuelos duró cada motor en total
max_ciclos = df.groupby('id_motor')['ciclo'].max().reset_index()
max_ciclos.columns = ['id_motor', 'max_ciclo']

# Cruzamos ese dato con la tabla original y restamos el ciclo actual
df = df.merge(max_ciclos, on='id_motor', how='left')
df['RUL'] = df['max_ciclo'] - df['ciclo']

# Borramos la columna auxiliar 'max_ciclo'
df = df.drop(columns=['max_ciclo'])


# 2. ELIMINAR SENSORES CONSTANTES (RUIDO)
# Si la desviación estándar (std) es 0, el valor nunca ha cambiado
desviaciones = df.std()
columnas_constantes = desviaciones[desviaciones == 0].index

# Eliminamos esas columnas de nuestro DataFrame
df = df.drop(columns=columnas_constantes)


# 3. MOSTRAR EL RESULTADO DE LA LIMPIEZA
print("Sensores eliminados por ser constantes:", list(columnas_constantes))
print("\nTotal de columnas útiles restantes:", len(df.columns))
display(df.head())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import numpy as np

# 1. SEPARAR CARACTERÍSTICAS (X) Y OBJETIVO (y)
X = df.drop(columns=['id_motor', 'ciclo', 'RUL'])
y = df['RUL']

# 2. NORMALIZAR LOS SENSORES
scaler = StandardScaler()
X_escalado = scaler.fit_transform(X)

# 3. SEPARAR PARA EL EXAMEN (Train y Test)
X_train, X_test, y_train, y_test = train_test_split(X_escalado, y, test_size=0.2, random_state=42)

# 4. ENTRENAR EL MODELO
print("Entrenando el modelo RandomForest (esto puede tardar unos 10-15 segundos)...")
modelo_rf = RandomForestRegressor(n_estimators=100, random_state=42)
modelo_rf.fit(X_train, y_train)

# 5. EVALUAR EL MODELO (Sintaxis actualizada)
predicciones = modelo_rf.predict(X_test)

# Calculamos el error cuadrático normal, y luego le hacemos la raíz cuadrada con Numpy
error_mse = mean_squared_error(y_test, predicciones)
error_rmse = np.sqrt(error_mse)

print(f"\nEntrenamiento completado.")
print(f"Error RMSE: El modelo se equivoca en promedio por +/- {error_rmse:.2f} vuelos.")
```

```python
# Hacemos una copia de la tabla limpia que ya teníamos
df_avanzado = df.copy()

# Lista de los sensores que nos quedan tras limpiar los constantes
# (Cogemos las columnas desde 'ajuste_1' en adelante, omitiendo el RUL que está al final)
columnas_sensores = df_avanzado.columns[2:-1]

print("Calculando promedios y tendencias acumuladas (Feature Engineering)...")

# Para cada motor independiente, calculamos tendencias usando ventanas de 10 vuelos
for motor in df_avanzado['id_motor'].unique():
    # Aislamos las filas de un solo motor
    filas_motor = df_avanzado['id_motor'] == motor

    # Aplicamos cálculos móviles (rolling) a las columnas de sensores de ese motor
    for col in columnas_sensores:
        # 1. Media de los últimos 10 vuelos (Tendencia suave)
        # min_periods=1 evita que las 9 primeras filas del motor se queden vacías
        df_avanzado.loc[filas_motor, f'{col}_media_10'] = df_avanzado.loc[filas_motor, col].rolling(window=10, min_periods=1).mean()

        # 2. Desviación de los últimos 10 vuelos (Inestabilidad y picos)
        # min_periods=2 porque la desviación necesita al menos 2 números para calcularse
        df_avanzado.loc[filas_motor, f'{col}_desv_10'] = df_avanzado.loc[filas_motor, col].rolling(window=10, min_periods=2).std().fillna(0)

print(f"Tamaño original: {df.shape}")
print(f"Nuevo tamaño tras añadir tendencias históricas: {df_avanzado.shape}")
display(df_avanzado.head())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import numpy as np

# 1. Separamos X (Sensores + Tendencias) e y (RUL) de nuestra nueva tabla ancha
X_avanzado = df_avanzado.drop(columns=['id_motor', 'ciclo', 'RUL'])
y_avanzado = df_avanzado['RUL']

# 2. Normalizamos todas las columnas (las originales y las de promedios)
scaler_av = StandardScaler()
X_avanzado_escalado = scaler_av.fit_transform(X_avanzado)

# 3. Separamos en entrenamiento y prueba (mismo 20% oculto que antes)
X_train_av, X_test_av, y_train_av, y_test_av = train_test_split(
    X_avanzado_escalado, y_avanzado, test_size=0.2, random_state=42
)

# 4. Entrenamos el modelo RandomForest
# (Puede tardar unos 20-30 segundos porque ahora procesa el triple de variables)
print("Entrenando RandomForest con Ingeniería de Características...")
modelo_rf_av = RandomForestRegressor(n_estimators=100, random_state=42)
modelo_rf_av.fit(X_train_av, y_train_av)

# 5. Evaluamos el modelo contra los datos de prueba
predicciones_av = modelo_rf_av.predict(X_test_av)
error_mse_av = mean_squared_error(y_test_av, predicciones_av)
error_rmse_av = np.sqrt(error_mse_av)

print(f"\nEntrenamiento completado.")
print(f"NUEVO Error RMSE: +/- {error_rmse_av:.2f} vuelos.")
print(f"Mejora respecto al modelo anterior (41.48): {41.48 - error_rmse_av:.2f} vuelos menos de error.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
import numpy as np

# Asumimos que 'df_avanzado' ya existe de los pasos anteriores.

# 1. IDENTIFICAR GRUPOS (Tu lógica)
# Sacamos los IDs únicos de los motores
motores = df_avanzado['id_motor'].unique()
# Dividimos los MÓTORES, no las filas sueltas
motores_train, motores_test = train_test_split(motores, test_size=0.2, random_state=42)

# 2. SEPARAR DATAFRAMES POR GRUPOS (Tu lógica)
# Filtramos la tabla completa para aislar los motores de train y de test
train_df = df_avanzado[df_avanzado['id_motor'].isin(motores_train)]
test_df = df_avanzado[df_avanzado['id_motor'].isin(motores_test)]

# 3. SEPARAR CARACTERÍSTICAS (X) Y OBJETIVO (y)
# Quitamos los IDs y la respuesta (RUL)
X_train = train_df.drop(columns=['id_motor', 'ciclo', 'RUL'])
y_train = train_df['RUL']

X_test = test_df.drop(columns=['id_motor', 'ciclo', 'RUL'])
y_test = test_df['RUL']

# 4. NORMALIZAR CORRECTAMENTE (Tu lógica)
scaler = StandardScaler()
# El scaler solo 'aprende' los rangos de TRAIN
X_train_escalado = scaler.fit_transform(X_train)
# Y aplicamos esos rangos a TEST, sin dejar que test influya (evita fuga)
X_test_escalado = scaler.transform(X_test)

# 5. ENTRENAR EL MODELO FINAL
print(f"Entrenando RandomForest SIN Fuga de Datos (motores train: {len(motores_train)}, motores test: {len(motores_test)})...")
modelo_rf_final = RandomForestRegressor(n_estimators=100, random_state=42)
modelo_rf_final.fit(X_train_escalado, y_train)

# 6. EVALUAR
predicciones_final = modelo_rf_final.predict(X_test_escalado)
error_mse_final = mean_squared_error(y_test, predicciones_final)
error_rmse_final = np.sqrt(error_mse_final)

print(f"\nEntrenamiento completado.")
print(f"Error RMSE REAL (legítimo): +/- {error_rmse_final:.2f} vuelos.")
```

```python
# 1. IMPORTACIONES NECESARIAS (Añadimos Pipeline)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import numpy as np

# Asumimos que 'df_avanzado' ya existe de los pasos anteriores.

# 2. SEPARACIÓN ESTRICTA POR GRUPOS (Tu lógica de motores)
# Identificamos los IDs únicos de los motores
motores = df_avanzado['id_motor'].unique()
# Barajamos y dividimos solo los IDs
motores_train, motores_test = train_test_split(motores, test_size=0.2, random_state=42)

# Filtramos la tabla avanzada usando .isin() para crear TRAIN y TEST aislados
train_df = df_avanzado[df_avanzado['id_motor'].isin(motores_train)]
test_df = df_avanzado[df_avanzado['id_motor'].isin(motores_test)]

# Separamos Características (X) y Objetivo (y)
X_train = train_df.drop(columns=['id_motor', 'ciclo', 'RUL'])
y_train = train_df['RUL']
X_test = test_df.drop(columns=['id_motor', 'ciclo', 'RUL'])
y_test = test_df['RUL']


# 3. DEFINICIÓN DEL PIPELINE (Paso a prueba de balas)
# Empaquetamos el escalador y el regresor. Scikit-Learn gestionará los fit/transform.
# 'Pasos' es una lista de tuplas: (nombre, objeto)
pasos_pipeline = [
    ('escalador', StandardScaler()),
    ('regresor', RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)) # n_jobs=-1 usa todos los núcleos para ir rápido
]

pipeline_mantenimiento = Pipeline(pasos_pipeline)


# 4. ENTRENAMIENTO Y EVALUACIÓN AUTOMÁTICA
print(f"Entrenando el Pipeline profesional (Motores train: {len(motores_train)})...")

# AL HACER FIT EN EL PIPELINE:
# 1. Llama a 'fit_transform()' del StandardScaler solo con X_train.
# 2. Llama a 'fit()' del RandomForestRegressor con X_train ya escalado.
pipeline_mantenimiento.fit(X_train, y_train)

print("Entrenamiento completado. Evaluando contra motores nuevos...")

# AL HACER PREDICT EN EL PIPELINE:
# 1. Llama a 'transform()' del StandardScaler usando X_test.
#    (Usa las medias/stds APRENDIDAS en TRAIN, garantizando CERO fuga de datos).
# 2. Llama a 'predict()' del RandomForestRegressor con X_test ya escalado.
predicciones_final = pipeline_mantenimiento.predict(X_test)

# Cálculo del error final legítimo
error_mse_pipe = mean_squared_error(y_test, predicciones_final)
error_rmse_pipe = np.sqrt(error_mse_pipe)

print(f"\nResultado Final del Pipeline Profesional:")
print(f"Error RMSE REAL (legítimo): +/- {error_rmse_pipe:.2f} vuelos.")
```

```python
# 1. IMPORTACIONES ADICIONALES (Añadimos XGBoost y Linear Models)
from xgboost import XGBRegressor
from sklearn.linear_model import Ridge
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import numpy as np
import pandas as pd # Para la tabla final

# Asumimos que X_train, X_test, y_train, y_test ya existen en memoria (sin fuga de datos, divididos por motores).

print("Iniciando la comparativa de modelos profesional...")

# 2. DEFINICIÓN DE LOS PIPELINES DE REFERENCIA
# Usamos exactamente la misma estructura de escalado para todos.

# Modelo 1: 'Suelo' (Ridge Regression - Regresión Lineal Regularizada)
# n_jobs=-1 usa todos los núcleos. Ridge es rápido y sirve de referencia básica.
pipe_ridge = Pipeline([
    ('escalador', StandardScaler()),
    ('regresor', Ridge(random_state=42)) # Ridge no usa n_jobs, pero es muy rápido
])

# Modelo 2: RandomForest (El que ya teníamos de +/- 35.09 flights)
pipe_rf = Pipeline([
    ('escalador', StandardScaler()),
    ('regresor', RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1))
])

# Modelo 3: XGBoost (Extremadamente potente, a menudo ganador en Kaggle)
pipe_xgb = Pipeline([
    ('escalador', StandardScaler()),
    # n_jobs=-1 usa todos los núcleos. Usamos parámetros base similares.
    ('regresor', XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42, n_jobs=-1))
])

# Diccionario para automatizar el entrenamiento y evaluación
modelos_a_comparar = {
    "1. Ridge (Lineal Regularizado)": pipe_ridge,
    "2. Random Forest": pipe_rf,
    "3. XGBoost Regressor": pipe_xgb
}

resultados = []

# 3. BUCLE AUTOMÁTICO DE ENTRENAMIENTO Y EVALUACIÓN
for nombre, pipeline in modelos_a_comparar.items():
    print(f"\nEntrenando {nombre}...")
    # El Pipeline gestiona automáticamente fit_transform(train)
    pipeline.fit(X_train, y_train)

    # El Pipeline gestiona automáticamente transform(test) sin fuga
    predicciones = pipeline.predict(X_test)

    # Cálculo de métricas
    error_rmse = np.sqrt(mean_squared_error(y_test, predicciones))
    resultados.append({"Modelo": nombre, "RMSE Real (Legítimo)": round(error_rmse, 2)})
    print(f"Finalizado {nombre}. RMSE: {error_rmse:.2f}")


# 4. TABLA DE REFERENCIA FINAL PARA TU PORTFOLIO
df_comparativa = pd.DataFrame(resultados).sort_values(by="RMSE Real (Legítimo)")

print("\n" + "="*50)
print("TABLA COMPARATIVA FINAL DE MODELOS")
print("="*50)
display(df_comparativa)
print("="*50)

# Recomendación automática del mejor modelo
mejor_modelo_nombre = df_comparativa.iloc[0]['Modelo']
print(f"Mejor modelo detectado (menor RMSE): {mejor_modelo_nombre}")
```

```python
import shap
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# ==========================================
# PARTE A: IMPORTANCIA DE SENSORES CON SHAP
# ==========================================
print("Calculando valores SHAP (esto puede tardar unos 10-20 segundos)...")
escalador = pipe_xgb.named_steps['escalador']
modelo_xgb = pipe_xgb.named_steps['regresor']

X_test_escalado = escalador.transform(X_test)
explainer = shap.TreeExplainer(modelo_xgb)
shap_values = explainer.shap_values(X_test_escalado)

# Gráfico SHAP (Summary Plot)
plt.figure(figsize=(12, 6))
shap.summary_plot(shap_values, X_test, plot_type="dot", show=False)
plt.title('A. Impacto de cada Sensor (Valores SHAP)', fontsize=14)
plt.tight_layout()
plt.show()

# ==========================================
# PARTE B: EVOLUCIÓN TEMPORAL DE LOS TOP 3
# ==========================================
# 1. Extraemos matemáticamente el Top 3 desde los valores SHAP
importancia_shap = np.abs(shap_values).mean(axis=0)
df_importancia = pd.DataFrame({
    'Variable': X_test.columns,
    'Importancia': importancia_shap
}).sort_values(by='Importancia', ascending=False)

top_3_sensores = df_importancia['Variable'].iloc[0:3].tolist()
print(f"\nLos 3 sensores que más delatan el fallo según SHAP son: {top_3_sensores}")

# 2. Dibujamos su evolución para el Motor 3
motor_ejemplo = df_avanzado[df_avanzado['id_motor'] == 3]

plt.figure(figsize=(12, 5))
for sensor in top_3_sensores:
    # Normalizamos (0 a 1) para poder comparar presiones y temperaturas en la misma escala
    sensor_norm = (motor_ejemplo[sensor] - motor_ejemplo[sensor].min()) / (motor_ejemplo[sensor].max() - motor_ejemplo[sensor].min())
    plt.plot(motor_ejemplo['ciclo'], sensor_norm, label=sensor, linewidth=2.5)

plt.title('B. Evolución Física de los Top 3 Sensores (Motor 3)', fontsize=14)
plt.xlabel('Ciclos de Vuelo (Edad del motor)')
plt.ylabel('Valor del Sensor (Normalizado 0-1)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

# Marcamos la Zona Crítica (últimos 50 vuelos)
ciclo_final = motor_ejemplo['ciclo'].max()
plt.axvspan(ciclo_final - 50, ciclo_final, color='red', alpha=0.1, label='Zona Crítica (<50 vuelos)')

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform

print("Iniciando la búsqueda de hiperparámetros para XGBoost...")

# 1. Definimos el espacio de búsqueda
# El prefijo 'regresor__' es obligatorio para que el Pipeline sepa a qué paso aplicar el parámetro
param_distributions = {
    'regresor__n_estimators': [100, 200, 300, 400],           # Cantidad de árboles
    'regresor__learning_rate': [0.01, 0.05, 0.1, 0.2],        # Velocidad de aprendizaje
    'regresor__max_depth': [3, 4, 5, 6, 7],                   # Profundidad de cada árbol (evita sobreajuste)
    'regresor__subsample': [0.7, 0.8, 0.9, 1.0],              # % de filas usadas en cada árbol
    'regresor__colsample_bytree': [0.7, 0.8, 0.9, 1.0]        # % de columnas usadas en cada árbol
}

# 2. Configuramos el buscador aleatorio
# cv=3 hace validación cruzada dividiendo el train en 3 trozos
# n_iter=20 probará 20 combinaciones aleatorias de nuestro espacio
random_search = RandomizedSearchCV(
    estimator=pipe_xgb,
    param_distributions=param_distributions,
    n_iter=20,
    scoring='neg_root_mean_squared_error',
    cv=3,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

# 3. Entrenamos (AQUÍ ES DONDE TARDA)
random_search.fit(X_train, y_train)

# 4. Extraemos el modelo ganador y sus parámetros
mejor_modelo_xgb = random_search.best_estimator_
mejores_parametros = random_search.best_params_

print("\n--- ¡BÚSQUEDA COMPLETADA! ---")
print("Mejores parámetros encontrados:")
for param, valor in mejores_parametros.items():
    print(f" - {param.replace('regresor__', '')}: {valor}")

# 5. Evaluamos el mejor modelo contra los datos de TEST sin ver
predicciones_optimizadas = mejor_modelo_xgb.predict(X_test)
rmse_optimizado = np.sqrt(mean_squared_error(y_test, predicciones_optimizadas))

print(f"\nRMSE del modelo optimizado: +/- {rmse_optimizado:.2f} vuelos.")
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# 1. Creamos una copia de trabajo para la clasificación
df_clasificacion = df_avanzado.copy()

# 2. Definimos la función de etiquetado por estados (Semáforo)
# RUL > 50: Sano (0)
# 15 < RUL <= 50: Alerta (1)
# RUL <= 15: Crítico (2 - Zona de peligro inminente)
def etiquetar_estado(rul):
    if rul > 50:
        return 0  # Sano
    elif rul > 15:
        return 1  # Alerta
    else:
        return 2  # Crítico

df_clasificacion['estado'] = df_clasificacion['RUL'].apply(etiquetar_estado)

# 3. Separamos X e y manteniendo tu división estricta por motores (usamos motores_train y motores_test anteriores)
train_class_df = df_clasificacion[df_clasificacion['id_motor'].isin(motores_train)]
test_class_df = df_clasificacion[df_clasificacion['id_motor'].isin(motores_test)]

X_train_c = train_class_df.drop(columns=['id_motor', 'ciclo', 'RUL', 'estado'])
y_train_c = train_class_df['estado']
X_test_c = test_class_df.drop(columns=['id_motor', 'ciclo', 'RUL', 'estado'])
y_test_c = test_class_df['estado']

# 4. Entrenamos un Pipeline de Clasificación con XGBoost
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

pipe_clasificador = Pipeline([
    ('escalador', StandardScaler()),
    ('clasificador', XGBClassifier(n_estimators=150, max_depth=3, learning_rate=0.05, random_state=42, n_jobs=-1))
])

print("Entrenando el clasificador de estado de salud...")
pipe_clasificador.fit(X_train_c, y_train_c)

# 5. Evaluamos con los datos de test
predicciones_clase = pipe_clasificador.predict(X_test_c)

# 6. Resultados y Matriz de Confusión
print("\n--- REPORTE DE CLASIFICACIÓN ---")
print(classification_report(y_test_c, predicciones_clase, target_names=['Sano (0)', 'Alerta (1)', 'Crítico (2)']))

# Dibujar la matriz de confusión
plt.figure(figsize=(6, 5))
matriz = confusion_matrix(y_test_c, predicciones_clase)
sns.heatmap(matriz, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Sano', 'Alerta', 'Crítico'],
            yticklabels=['Sano', 'Alerta', 'Crítico'])
plt.title('Matriz de Confusión Operacional')
plt.xlabel('Predicción del Modelo')
plt.ylabel('Realidad Física')
plt.tight_layout()
plt.show()
```
